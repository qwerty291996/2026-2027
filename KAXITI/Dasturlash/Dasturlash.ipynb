## Savol 1. Boʻlakli funksiya qiymatini hisoblash

*a* haqiqiy son berilgan. Quyidagi rasmda tasvirlangan y(x) funksiya uchun y(*a*) ni hisoblang.



Rasmdagi belgilarga koʻra funksiya uchta boʻlakdan iborat (sinish nuqtalari **−1** va **2**):

$$
y(x)=\begin{cases} -\dfrac{1}{x^{2}}, & x<-1\\[6pt] x^{2}, & -1\le x\le 2\\[6pt] x, & x>2 \end{cases}
$$

---

### Yechim gʻoyasi

Bu yerda yagona formula yoʻq: *x* qaysi oraliqda yotishiga qarab tegishli formulani tanlaymiz. Buning uchun **shart operatori** (agar–aks holda) ishlatiladi. Uchta oraliq — uchta holat:

- *x* < −1 boʻlsa, javob −1/x²;
- *x* −1 bilan 2 orasida (chegaralar bilan) boʻlsa, javob x²;
- aks holda (*x* > 2), javob x.

---

### Algoritm

**1-usul. Tabiiy til (soʻz bilan)**

1. *a* sonini kiritamiz.
2. Agar *a* < −1 boʻlsa, javob = −1/*a*².
3. Aks holda, agar *a* ≤ 2 boʻlsa, javob = *a*².
4. Aks holda, javob = *a*.
5. Javobni chiqaramiz.

**2-usul. Psevdokod**

```
Algoritm: Y(a)
Kirish:  haqiqiy son a
Chiqish: y(a) qiymati

1. agar a < -1 boʻlsa:
2.     natija ← -1 / (a * a)
3. aks holda agar a ≤ 2 boʻlsa:
4.     natija ← a * a
5. aks holda:
6.     natija ← a
7. natija ni qaytar
```

### C++ kodi — avval faylga yozamiz

````python
%%writefile y_func.cpp
#include <iostream>
using namespace std;

// Rasmda berilgan boʻlakli funksiya:
//   y = -1/x^2 ,  agar x < -1
//   y =  x^2   ,  agar -1 <= x <= 2
//   y =  x     ,  agar x > 2
double y(double x) {
    if (x < -1) {
        return -1.0 / (x * x);   // -1/x^2
    } else if (x <= 2) {
        return x * x;            // x^2
    } else {
        return x;                // x
    }
}

int main() {
    // Har uchala boʻlakni tekshirish uchun namuna qiymatlar
    double testlar[] = {-2, -0.5, 2, 3};
    for (double a : testlar) {
        cout << "y(" << a << ") = " << y(a) << endl;
    }
    return 0;
}
````

### Python kodi

````python
def y(x):
    # Rasmda berilgan boʻlakli funksiya
    if x < -1:
        return -1 / (x * x)   # -1/x^2
    elif x <= 2:
        return x * x          # x^2
    else:
        return x              # x


# Har uchala boʻlakni tekshirish uchun namuna qiymatlar
for a in [-2, -0.5, 2, 3]:
    print(f"y({a}) = {y(a)}")
````

Har bir namuna funksiyaning bir boʻlagini tekshiradi: *a* = −2 → birinchi boʻlak (−1/x²), *a* = −0.5 va *a* = 2 → oʻrtadagi boʻlak (x²), *a* = 3 → oxirgi boʻlak (x).
