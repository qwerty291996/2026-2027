## Saralash algoritmlari

**Saralash** — bu berilgan sonlar roʻyxatini maʼlum tartibda (masalan, kichikdan kattaga qarab) qayta joylashtirish. Saralangan roʻyxatda kerakli sonni topish, eng katta yoki eng kichigini aniqlash, takrorlarni sezish — barchasi ancha osonlashadi. Shu sababli saralash — dasturlashning eng koʻp uchraydigan asosiy amallaridan biri.

Quyida eng sodda uch usulni koʻrib chiqamiz. Uchalasi bir xil natijaga olib keladi, ammo *fikrlash yoʻli* har xil. Har birini bitta misolda — `[5, 2, 4, 1]` roʻyxatini oʻsish tartibida saralash orqali — tushuntiramiz.


---

### 1. Pufaksimon saralash (bubble sort)

**Gʻoyasi.** Roʻyxat boʻylab yurib, **qoʻshni** ikki sonni solishtiramiz. Agar chapdagisi oʻngdagisidan katta boʻlsa, ularning oʻrnini almashtiramiz. Bir marta toʻliq yurib chiqilganda eng katta son roʻyxatning oxiriga "suzib" chiqadi. Buni roʻyxat toʻliq tartiblanmaguncha takrorlaymiz. Nomi shundan: katta sonlar suvdagi pufakdek yuqoriga koʻtariladi.

**Algoritm**

**1-usul. Tabiiy til (soʻz bilan)**

1. Roʻyxat *a* va uning uzunligi *n* ni olamiz.
2. Roʻyxat boʻylab bir necha marta yuramiz (jami *n*−1 marta yetadi).
3. Har yurishda birinchi juftdan boshlab qoʻshni sonlarni solishtiramiz: agar `a[j] > a[j+1]` boʻlsa, oʻrnini almashtiramiz.
4. Barcha yurishlar tugagach, roʻyxat saralangan boʻladi.

**2-usul. Psevdokod**

```
Algoritm: Pufaksimon(a, n)
Kirish:  a ro'yxat, n uzunlik
Chiqish: o'sish tartibida saralangan a

1. i ← 0 dan n-2 gacha takrorla:
2.     j ← 0 dan n-2-i gacha takrorla:
3.         agar a[j] > a[j+1] bo'lsa:
4.             a[j] va a[j+1] o'rnini almashtir
5. a ni qaytar
```

**3-usul. Blok-sxema**



**Qadamlarni kuzatamiz** — `[5, 2, 4, 1]`:

```
1-yurish: (5,2)→almash [2,5,4,1] → (5,4)→almash [2,4,5,1] → (5,1)→almash [2,4,1,5]
2-yurish: (2,4) joyida → (4,1)→almash [2,1,4,5] → (4,5) joyida
3-yurish: (2,1)→almash [1,2,4,5]
Natija:   [1, 2, 4, 5]
```

**C++ kodi — avval faylga yozamiz**

````python
%%writefile pufaksimon.cpp
#include <iostream>
using namespace std;

// Pufaksimon saralash: qo'shni juftlarni solishtirib, almashtiramiz
void pufaksimon(int a[], int n) {
    for (int i = 0; i < n - 1; i++) {          // nechta to'liq yurish
        for (int j = 0; j < n - 1 - i; j++) {  // qo'shni juftlar
            if (a[j] > a[j + 1]) {             // chapdagi kattaroqmi?
                int vaqt = a[j];               // o'rnini almashtiramiz
                a[j] = a[j + 1];
                a[j + 1] = vaqt;
            }
        }
    }
}

int main() {
    int a[] = {5, 2, 4, 1};
    int n = 4;
    pufaksimon(a, n);
    cout << "Saralangan: ";
    for (int i = 0; i < n; i++) cout << a[i] << " ";
    cout << endl;
    return 0;
}
````

**Endi kompilyatsiya qilib ishga tushiramiz**

````python
!g++ pufaksimon.cpp -o pufaksimon.exe && pufaksimon.exe
````

**Python kodi**


````python
def pufaksimon(a):
    n = len(a)
    for i in range(n - 1):              # nechta to'liq yurish
        for j in range(n - 1 - i):      # qo'shni juftlar
            if a[j] > a[j + 1]:         # chapdagi kattaroqmi?
                a[j], a[j + 1] = a[j + 1], a[j]   # o'rnini almashtiramiz
    return a


a = [5, 2, 4, 1]
print("Saralangan:", pufaksimon(a))
````

### 2. Tanlab saralash (selection sort)

**Gʻoyasi.** Roʻyxatni ikki qismga boʻlib tasavvur qilamiz: chapda saralangan qism, oʻngda saralanmagani. Har qadamda saralanmagan qismdan eng kichik sonni topamiz va uni oʻsha qismning eng boshiga qoʻyamiz. Shunda saralangan qism har qadamda bittaga uzayadi.

**Algoritm**

**1-usul. Soʻz bilan**

1. Roʻyxat *a* va uzunlik *n* ni olamiz.
2. Har bir *i*-oʻrin uchun (0 dan *n*−2 gacha): *i* dan oxirigacha boʻlgan qismdan eng kichik sonning oʻrnini topamiz.
3. Shu eng kichik sonni *i*-oʻrindagi son bilan almashtiramiz.
4. Barcha oʻrinlar toʻlgach, roʻyxat saralangan boʻladi.

**2-usul. Psevdokod**

```
Algoritm: Tanlab(a, n)
Kirish:  a ro'yxat, n uzunlik
Chiqish: o'sish tartibida saralangan a

1. i ← 0 dan n-2 gacha takrorla:
2.     mi ← i                       (eng kichigi indeksi)
3.     j ← i+1 dan n-1 gacha takrorla:
4.         agar a[j] < a[mi] bo'lsa:
5.             mi ← j
6.     a[i] va a[mi] o'rnini almashtir
7. a ni qaytar
```

**3-usul. Blok-sxema**

**Qadamlarni kuzatamiz** — `[5, 2, 4, 1]`:

```
1-qadam: [5,2,4,1] ichida eng kichik = 1 → 5 bilan almash → [1, 2, 4, 5]
2-qadam: [2,4,5]   ichida eng kichik = 2 → joyida     → [1, 2, 4, 5]
3-qadam: [4,5]     ichida eng kichik = 4 → joyida     → [1, 2, 4, 5]
Natija:  [1, 2, 4, 5]
```

**C++ kodi — avval faylga yozamiz**

````python
%%writefile tanlab.cpp
#include <iostream>
using namespace std;

// Tanlab saralash: har safar eng kichigini topib, oldinga qo'yamiz
void tanlab(int a[], int n) {
    for (int i = 0; i < n - 1; i++) {
        int min_idx = i;                      // eng kichigi shu deb faraz qilamiz
        for (int j = i + 1; j < n; j++) {
            if (a[j] < a[min_idx]) {          // undan kichigini topsak
                min_idx = j;                  // indeksini yodda tutamiz
            }
        }
        int vaqt = a[i];                      // eng kichigini i-o'ringa qo'yamiz
        a[i] = a[min_idx];
        a[min_idx] = vaqt;
    }
}

int main() {
    int a[] = {5, 2, 4, 1};
    int n = 4;
    tanlab(a, n);
    cout << "Saralangan: ";
    for (int i = 0; i < n; i++) cout << a[i] << " ";
    cout << endl;
    return 0;
}
````

**Endi kompilyatsiya qilib ishga tushiramiz**

````python
!g++ tanlab.cpp -o tanlab.exe && tanlab.exe
````

**Python kodi**

````python
def tanlab(a):
    n = len(a)
    for i in range(n - 1):
        min_idx = i                      # eng kichigi shu deb faraz qilamiz
        for j in range(i + 1, n):
            if a[j] < a[min_idx]:        # undan kichigini topsak
                min_idx = j              # indeksini yodda tutamiz
        a[i], a[min_idx] = a[min_idx], a[i]   # eng kichigini oldinga qo'yamiz
    return a


a = [5, 2, 4, 1]
print("Saralangan:", tanlab(a))
````

### 3. Joylashtirib saralash (insertion sort)

**Gʻoyasi.** Xuddi qoʻlda karta tartiblashga oʻxshaydi. Chapdagi saralangan qatorga oʻngdan bittalab son olib kelamiz va uni saralangan qismda **oʻz oʻrniga** suqib qoʻyamiz: undan kattalarni bittaga oʻngga surib, boʻshagan joyga qoʻyamiz.

**Algoritm**

**1-usul. Tabiiy til (soʻz bilan)**

1. Roʻyxat *a* va uzunlik *n* ni olamiz. Birinchi son (bitta oʻzi) saralangan hisoblanadi.
2. Ikkinchi sondan boshlab har bir sonni "kalit" deb olamiz.
3. Kalitdan chap tomondagi undan katta sonlarni bittaga oʻngga suramiz.
4. Boʻshagan joyga kalitni qoʻyamiz.
5. Roʻyxat oxiriga yetganda, u toʻliq saralangan boʻladi.

**2-usul. Psevdokod**

```
Algoritm: Joylashtirib(a, n)
Kirish:  a ro'yxat, n uzunlik
Chiqish: o'sish tartibida saralangan a

1. i ← 1 dan n-1 gacha takrorla:
2.     kalit ← a[i]
3.     j ← i-1
4.     j ≥ 0 va a[j] > kalit bo'lguncha takrorla:
5.         a[j+1] ← a[j]        (o'ngga suramiz)
6.         j ← j-1
7.     a[j+1] ← kalit           (o'z o'rniga qo'yamiz)
8. a ni qaytar
```

**3-usul. Blok-sxema**


**Qadamlarni kuzatamiz** — `[5, 2, 4, 1]`:

```
Boshlang'ich: [5 | 2, 4, 1]   (birinchi son saralangan)
2 ni qo'yamiz: 2 < 5           → [2, 5 | 4, 1]
4 ni qo'yamiz: 4 < 5, 4 > 2    → [2, 4, 5 | 1]
1 ni qo'yamiz: 1 < 5,4,2       → [1, 2, 4, 5]
Natija:        [1, 2, 4, 5]
```

**C++ kodi — avval faylga yozamiz**

````python
%%writefile joylashtirib.cpp
#include <iostream>
using namespace std;

// # Joylashtirib saralash: har bir sonni saralangan qismga o'z o'rniga qo'yamiz
void joylashtirib(int a[], int n) {
    for (int i = 1; i < n; i++) {
        int kalit = a[i];               // joylashtiriladigan son
        int j = i - 1;
        // # kalitdan katta bo'lganlarni bittaga o'ngga suramiz
        while (j >= 0 && a[j] > kalit) {
            a[j + 1] = a[j];
            j = j - 1;
        }
        a[j + 1] = kalit;               // # kalitni bo'shagan joyga qo'yamiz
    }
}

int main() {
    int a[] = {5, 2, 4, 1};
    int n = 4;
    joylashtirib(a, n);
    cout << "Saralangan: ";
    for (int i = 0; i < n; i++) cout << a[i] << " ";
    cout << endl;
    return 0;
}
````

**Endi kompilyatsiya qilib ishga tushiramiz**


````python
!g++ joylashtirib.cpp -o joylashtirib.exe && joylashtirib.exe
````

**Python kodi**

````python
def joylashtirib(a):
    n = len(a)
    for i in range(1, n):
        kalit = a[i]                     # joylashtiriladigan son
        j = i - 1
        # kalitdan katta bo'lganlarni bittaga o'ngga suramiz
        while j >= 0 and a[j] > kalit:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = kalit                 # kalitni bo'shagan joyga qo'yamiz
    return a


a = [5, 2, 4, 1]
print("Saralangan:", joylashtirib(a))
````

### Uchala usulni taqqoslash

Uchala algoritm ham bir xil saralangan roʻyxatni beradi, farqi — ishlash uslubida:

- **pufaksimon** — qoʻshnilarni qayta-qayta almashtiradi; tushunish eng oson, lekin koʻp almashtirish qiladi;
- **tanlab** — har safar eng kichigini qidiradi va oldinga qoʻyadi; almashtirishlar soni kam;
- **joylashtirib** — har bir sonni tartibli qismga suqib qoʻyadi; roʻyxat deyarli tartibli boʻlsa juda tez ishlaydi.

Uchalasi ham kichik roʻyxatlar uchun juda qulay. Roʻyxat kattalashganda (masalan minglab son) ular sekinlashadi — bunday holatlar uchun tezroq algoritmlar bor, lekin ularning gʻoyasini tushunish uchun avval mana shu uch sodda usulni bilish kerak.
