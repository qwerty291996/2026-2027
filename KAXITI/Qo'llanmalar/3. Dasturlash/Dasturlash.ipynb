## Savol 1. *a* haqiqiy son berilgan. Quyidagi rasmda tasvirlangan y(x) funksiya uchun y(*a*) ni hisoblang.



Rasmdagi belgilarga koʻra funksiya uchta boʻlakdan iborat (sinish nuqtalari **−1** va **2**):

$$
y(x)=\begin{cases} -\dfrac{1}{x^{2}}, & x<-1\\[6pt] x^{2}, & -1\le x\le 2\\[6pt] x, & x>2 \end{cases}
$$

---

### Yechim gʻoyasi

Bu yerda yagona formula yoʻq: *x* qaysi oraliqda yotishiga qarab tegishli formulani tanlaymiz. Buning uchun **shart operatori** (if–else) ishlatiladi. Uchta oraliq — uchta holat:

- *x* < −1 boʻlsa, javob −1/x²;
- *x* −1 bilan 2 orasida (chegaralar bilan) boʻlsa, javob x²;
- aks holda (*x* > 2), javob x.

---

### Algoritm

**1-usul. Tabiiy til (soʻz bilan)**

1. *a* sonini kiritamiz.
2. Agar *a* < −1 boʻlsa, javob = −1/*a*².
3. Aks holda, agar *a* ≤ 2 boʻlsa, javob = *a*².
4. Aks holda, javob = *a*.
5. Javobni chiqaramiz.

**2-usul. Psevdokod**

```
Algoritm: Y(a)
Kirish:  haqiqiy son a
Chiqish: y(a) qiymati

1. agar a < -1 boʻlsa:
2.     natija ← -1 / (a * a)
3. aks holda agar a ≤ 2 boʻlsa:
4.     natija ← a * a
5. aks holda:
6.     natija ← a
7. natija ni qaytar
```

````python
%%writefile y_func.cpp
#include <iostream>
using namespace std;

// Rasmda berilgan boʻlakli funksiya:
//   y = -1/x^2 ,  agar x < -1
//   y =  x^2   ,  agar -1 <= x <= 2
//   y =  x     ,  agar x > 2
double y(double x) {
    if (x < -1) {
        return -1.0 / (x * x);   // -1/x^2
    } else if (x <= 2) {
        return x * x;            // x^2
    } else {
        return x;                // x
    }
}

int main() {
    // Har uchala boʻlakni tekshirish uchun namuna qiymatlar
    double testlar[] = {-2, -0.5, 2, 3};
    for (double a : testlar) {
        cout << "y(" << a << ") = " << y(a) << endl;
    }
    return 0;
}
````

## Savol 2. *x*, *y* haqiqiy sonlar berilgan. Koordinatalari (*x*, *y*) boʻlgan nuqta quyida keltirilgan rasmdagi tekislikning shtrixlangan qismiga tegishlimi? (ha/yoʻq)


---

### Yechim gʻoyasi

Bunday masalada asosiy ish — shtrixlangan sohani **tengsizliklar** bilan yozib olish. Nuqta shu tengsizliklarning barchasini bir vaqtda qanoatlantirsa, u sohaga tegishli.

Rasmni oʻqiymiz:

- markazdan chiquvchi ikki chiziq (−1, 1) va (1, 1) nuqtalardan oʻtadi — bu `y = |x|` (chapda `y = −x`, oʻngda `y = x`);
- yuqori gorizontal chegara — `y = 1`;
- pastki chegara — `y = 0` (abssissa oʻqi).

Oq (shtrixlanmagan) qism — markazdagi V shaklidagi uchburchak: `|x| < y < 1`. Undan tashqaridagi butun yuqori yarim tekislik shtrixlangan. Demak nuqta **shtrixlangan sohaga tegishli** boʻlish sharti:

$$
y \ge 0 \quad\text{va}\quad \big(\,y \ge 1 \ \text{ yoki } \ |x| \ge y\,\big)
$$

---

### Algoritm

**1-usul. Tabiiy til (soʻz bilan)**

1. *x* va *y* ni kiritamiz.
2. Agar *y* < 0 boʻlsa — nuqta pastki yarim tekislikda, javob **yoʻq**.
3. Aks holda, agar *y* ≥ 1 boʻlsa — nuqta yuqori tasmada, javob **ha**.
4. Aks holda, agar |*x*| ≥ *y* boʻlsa — nuqta yon yelkada, javob **ha**.
5. Aks holda — nuqta markazdagi oq uchburchakda, javob **yoʻq**.

**2-usul. Psevdokod**

```
Algoritm: Tegishli(x, y)
Kirish:  haqiqiy sonlar x, y
Chiqish: "ha" yoki "yoʻq"

1. agar (y ≥ 0) va (y ≥ 1 yoki |x| ≥ y) boʻlsa:
2.     "ha" ni qaytar
3. aks holda:
4.     "yoʻq" ni qaytar
```

### C++ kodi — avval faylga yozamiz

````python
%%writefile soha.cpp
#include <iostream>
#include <cmath>
using namespace std;

// Nuqta shtrixlangan sohaga tegishlimi?
// Soha:  y >= 0  VA  ( y >= 1  YOKI  |x| >= y )
bool sohaga_tegishli(double x, double y) {
    return (y >= 0) && (y >= 1 || fabs(x) >= y);
}

int main() {
    // Har bir qismni tekshirish uchun namuna nuqtalar
    double testlar[][2] = {{0, 0.5}, {0.9, 0.3}, {0, 2}, {0, -1}, {0.5, 0.9}};
    for (auto p : testlar) {
        double x = p[0], y = p[1];
        cout << "(" << x << ", " << y << ") -> "
             << (sohaga_tegishli(x, y) ? "ha" : "yo'q") << endl;
    }
    return 0;
}
````

## Savol 3. Berilgan *n* soni uchun quyidagi yigʻindini hisoblash algoritmi va dasturiy taʼminotini tuzing:

$$
S=\frac{x-1}{1}-\frac{(x-1)^{2}}{2}+\frac{(x-1)^{3}}{3}-\frac{(x-1)^{4}}{4}+\cdots+\frac{(-1)^{n-1}(x-1)^{n}}{n}
$$

---

### Yechim gʻoyasi

Yigʻindining har bir hadiga diqqat qilamiz. *k*-hadning umumiy koʻrinishi:

$$
\frac{(-1)^{k-1}(x-1)^{k}}{k}
$$

Bu yerda uch qism bor: **ishora** `(-1)^(k-1)` hadma-had almashadi (musbat, manfiy, musbat, …), **daraja** `(x-1)^k`, va **maxraj** `k`. Demak *k* ni 1 dan *n* gacha yurgizib, har bir hadni hisoblab, yigʻindiga qoʻshib boramiz.

Kirish: *x* (haqiqiy son) va *n* (natural son). Chiqish: *S*.

Bu turdagi masalada asosiy vosita — **takrorlash (sikl)**.

---

### Algoritm

**1-usul. Tabiiy til (soʻz bilan)**

1. *x* va *n* ni kiritamiz.
2. Yigʻindini nolga tenglaymiz: `S ← 0`.
3. *k* ni 1 dan *n* gacha bittalab oshirib boramiz. Har bir *k* uchun:
   - hadni hisoblaymiz: `(-1)^(k-1)·(x-1)^k / k`;
   - uni `S` ga qoʻshamiz.
4. Sikl tugagach, `S` ni chiqaramiz.

**2-usul. Psevdokod**

```
Algoritm: Yigindi(x, n)
Kirish:  haqiqiy son x, natural son n
Chiqish: S yigindisi

1. S ← 0
2. k ← 1 dan n gacha takrorla:
3.     ishora ← (-1)^(k-1)
4.     S ← S + ishora * (x-1)^k / k
5. S ni qaytar
```

**3-usul. Blok-sxema**



---

### C++ kodi — avval faylga yozamiz

````python
%%writefile yigindi.cpp
#include <iostream>
#include <cmath>
using namespace std;

// Yig'indi: S = sum_{k=1..n} (-1)^(k-1) * (x-1)^k / k
double yigindi(double x, int n) {
    double S = 0.0;
    for (int k = 1; k <= n; k++) {
        double ishora = (k % 2 == 1) ? 1.0 : -1.0;   // (-1)^(k-1)
        S += ishora * pow(x - 1, k) / k;
    }
    return S;
}

int main() {
    double x = 1.5;
    int n = 5;
    cout << "x = " << x << ", n = " << n << endl;
    cout << "S = " << yigindi(x, n) << endl;
    return 0;
}
````

### Python kodi

````python
def yigindi(x, n):
    # S = sum_{k=1..n} (-1)^(k-1) * (x-1)^k / k
    S = 0.0
    for k in range(1, n + 1):
        ishora = 1 if k % 2 == 1 else -1      # (-1)^(k-1)
        S += ishora * (x - 1) ** k / k
    return S


x = 1.5
n = 5
print(f"x = {x}, n = {n}")
print(f"S = {yigindi(x, n)}")
````

## Savol 4. Berilgan *n* soni uchun quyidagi yigʻindini hisoblash algoritmi va dasturiy taʼminotini tuzing:

$$

Kirish: *k* (haqiqiy son) va *n* (natural son). Chiqish: *S*.

### Algoritm

**1-usul. Tabiiy til (soʻz bilan)**

1. *k* va *n* ni kiritamiz.
2. Yigʻindini nolga tenglaymiz: `S ← 0`.
3. *i* ni 0 dan *n* gacha bittalab oshirib boramiz. Har bir *i* uchun:
   - hadni hisoblaymiz: `(-1)^i · k^i / i!`;

### C++ kodi — avval faylga yozamiz

````python
%%writefile yigindi4.cpp
#include <iostream>
#include <cmath>
using namespace std;

// i! (i faktorial) ni hisoblaydi
double faktorial(int m) {
    double f = 1.0;
    for (int i = 2; i <= m; i++) f *= i;
    return f;
}

// S = sum_{i=0..n} (-1)^i * k^i / i!
double yigindi(double k, int n) {
    double S = 0.0;
    for (int i = 0; i <= n; i++) {
        double ishora = (i % 2 == 0) ? 1.0 : -1.0;   // (-1)^i
        S += ishora * pow(k, i) / faktorial(i);
    }
    return S;
}

int main() {
    double k = 1.0;
    int n = 4;
    cout << "k = " << k << ", n = " << n << endl;
    cout << "S = " << yigindi(k, n) << endl;
    return 0;
}
````

### Python kodi

````python
from math import factorial

def yigindi(k, n):
    # S = sum_{i=0..n} (-1)^i * k^i / i!
    S = 0.0
    for i in range(0, n + 1):
        ishora = 1 if i % 2 == 0 else -1      # (-1)^i
        S += ishora * k ** i / factorial(i)
    return S


k = 1.0
n = 4
print(f"k = {k}, n = {n}")
print(f"S = {yigindi(k, n)}")
````

### Boshqa yechish usullari

Xuddi shu yigʻindini yana ikki koʻrinishda hisoblash mumkin. Uchalasi ham bir xil natija beradi.

**2-usul. Daraja va faktorialni bosqichma-bosqich.** Har safar `k^i` va `i!` ni noldan hisoblash oʻrniga, oldingi qiymatlarni yangilab boramiz (`daraja *= k`, `fakt *= i`). Bu ancha tezroq ishlaydi:

```python
def yigindi_2(k, n):
    S = 0.0
    daraja = 1.0    # k^0
    fakt = 1.0      # 0!
    ishora = 1      # birinchi had musbat
    for i in range(0, n + 1):
        if i > 0:
            daraja *= k     # daraja = k^i
            fakt *= i       # fakt = i!
        S += ishora * daraja / fakt
        ishora = -ishora
    return S
```

```cpp
double yigindi_2(double k, int n) {
    double S = 0.0, daraja = 1.0, fakt = 1.0, ishora = 1.0;
    for (int i = 0; i <= n; i++) {
        if (i > 0) { daraja *= k; fakt *= i; }   // daraja=k^i, fakt=i!
        S += ishora * daraja / fakt;
        ishora = -ishora;
    }
    return S;
}
```

**3-usul. Hadlar orasidagi bogʻlanish (rekurrent).** Ketma-ket hadlar bir-biriga bogʻliq: har bir yangi had oldingisidan `-k / i` koʻpaytuvchi orqali olinadi. Shunda na daraja, na faktorialni alohida saqlash shart:

```python
def yigindi_3(k, n):
    had = 1.0     # 0-had = 1
    S = 1.0
    for i in range(1, n + 1):
        had *= -k / i          # oldingi haddan keyingisi
        S += had
    return S
```

```cpp
double yigindi_3(double k, int n) {
    double had = 1.0, S = 1.0;   // 0-had = 1
    for (int i = 1; i <= n; i++) {
        had *= -k / i;           // oldingi haddan keyingisi
        S += had;
    }
    return S;
}
```

1-usul formulani harfma-harf takrorlaydi (eng tushunarli), 2-usul ortiqcha hisoblashni olib tashlaydi (tezroq), 3-usul esa hadlar orasidagi qonuniyatni ishlatadi (eng ixcham). Kichik izoh: *n* katta boʻlgani sari bu yigʻindi `e^(-k)` qiymatiga yaqinlashadi — bu `e^(-k)` funksiyasining qatorga yoyilmasidir.
